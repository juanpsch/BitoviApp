```python

from langfuse import get_client

langfuse = get_client()
DATASET_NAME = "intent_analyzer_structured_set"

try:
    langfuse.create_dataset(name=DATASET_NAME, description="Evaluación de intenciones con Structured Output")
except Exception:
    pass

# Mapeo exacto según tus categorías de TaskType
qs_mapeadas = [
    {"q": "Whats the last article about?", "intent": "synthesis"}, # Pide el contenido/resumen del artículo
    {"q": "List me the last post related to Inteligencia Artificial", "intent": "listing"}, # Pide una lista/apuntar a docs
    {"q": "List me all post related to RAG", "intent": "listing"}, # "List me all..." -> listing
    {"q": "list articles or testing of 2024", "intent": "listing"}, # "list..." -> listing
    {"q": "what is amazon's revenue in 2024?", "intent": "synthesis"}, # Extrayendo info/explicación fáctica
    {"q": "whats a vector store?", "intent": "synthesis"}, # "Qué es..." -> synthesis
    {"q": "Whats is the last article about?", "intent": "synthesis"}, 
    {"q": "Last article about?", "intent": "synthesis"},
    {"q": "Whats Retreival Augmnted Generation?", "intent": "synthesis"}, # "Qué es..." -> synthesis
    {"q": "whats the last article about?", "intent": "synthesis"},
    {"q": "Whats a Rag?", "intent": "synthesis"} # "Qué es..." -> synthesis
]

for item in qs_mapeadas:
    langfuse.create_dataset_item(
        dataset_name=DATASET_NAME,
        input={"messages": [("user", item["q"])]},
        expected_output={"intent": item["intent"]} # Esperamos el string del Enum ("listing" o "synthesis")
    )

print(f"✅ Dataset estructurado listo con {len(qs_mapeadas)} ítems.")
```
